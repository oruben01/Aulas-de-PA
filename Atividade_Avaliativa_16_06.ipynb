```python
# ---------------------------------------------------------
# QUESTÃO 02 – Soma dos Números (2 pontos)
# ---------------------------------------------------------

# Solicite ao usuário um número inteiro positivo.

# Utilizando um laço de repetição, calcule e exiba a soma de
# todos os números de 1 até o número informado.

# Exemplo:

# Digite um número: 5

# Resultado: 15

# Pois:

# 1 + 2 + 3 + 4 + 5 = 15

print("===========================================================")
a = int(input("Para começar, digite um número: "))

soma = 0

for i in range(1, a + 1):
    soma += i
print("-----------------------------------------------------------")
print(f"O resultado é: {soma}")
```

```python
# QUESTÃO 03 – Tabuada (2 pontos)
# ---------------------------------------------------------

# Solicite ao usuário um número inteiro e exiba sua tabuada
# de 1 até 10.

# Exemplo:

# Digite um número: 7

# 7 x 1 = 7
# 7 x 2 = 14
# 7 x 3 = 21
# ...
# 7 x 10 = 70

print("=======================================================================")
num = int(input("Digite um número: "))
print("-----------------------------------------------------------------------")
print(f"A tabuada do {num} é:")
print("-----------------------------------------------------------------------")

for i in range(1, 11):
    resultado = num * i

    print(f"{num} x {i} = {resultado}")

print("=====================================================================")
```

```python
# ---------------------------------------------------------
# QUESTÃO 05 – Caixa Eletrônico (5 pontos)
# ---------------------------------------------------------

# Um caixa eletrônico possui notas de:

# R$ 100
# R$ 50
# R$ 20
# R$ 10
# R$ 5
# R$ 2
# R$ 1

# Solicite ao usuário o valor do saque e informe a menor
# quantidade possível de notas para realizar o pagamento.

# Exemplo:

# Digite o valor do saque: 287

# Saída esperada:

# 2 nota(s) de R$100
# 1 nota(s) de R$50
# 1 nota(s) de R$20
# 1 nota(s) de R$10
# 1 nota(s) de R$5
# 1 nota(s) de R$2

# Requisitos:
# - Utilizar laços de repetição.
# - Não utilizar listas.
# - Exibir apenas as notas utilizadas.

print("===$===$===$===$===$===$===$===$===$===$===$===$===$===$===$===$===$===")
valor = int(input("Digite o valor do saque: "))
print("===$===$===$===$===$===$===$===$===$===$===$===$===$===$===$===$===$===")

notas100 = 0

while valor >= 100:
    notas100 += 1
    valor -= 100

if notas100 > 0:
    print(f"{notas100} nota(s) de R$100")

notas50 = 0
while valor >= 50:
    notas50 += 1
    valor -= 50
if notas50 > 0:
    print(f"{notas50} nota(s) de R$50")

notas20 = 0
while valor >= 20:
    notas20 += 1
    valor -= 20
if notas20 > 0:
    print(f"{notas20} nota(s) de R$20")

notas10 = 0
while valor >= 10:
    notas10 += 1
    valor -= 10
if notas10 > 0:
    print(f"{notas10} nota(s) de R$10")

notas5 = 0
while valor >= 5:
    notas5 += 1
    valor -= 5
if notas5 > 0:
    print(f"{notas5} nota(s) de R$5")

notas2 = 0
while valor >= 2:
    notas2 += 1
    valor -= 2
if notas2 > 0:
    print(f"{notas2} nota(s) de R$2")

notas1 = 0
while valor >= 1:
    notas1 += 1
    valor -= 1
if notas1 > 0:
    print(f"{notas1} nota(s) de R$1")

print("----------------------------------------------------------------------")
```
