```python
# Jogo Menu inteligente

# Criar um sistema que funcione até o usuário escolher sair.

# Menu

# 1 - Par/ Impar
# 2 - Positivo / Negativo
# 3 - Maior
# 4 - Sair

opcao = ""

try:

  while opcao != 4:

    print("--- Menu ---")
    print("1 - Par/ Impar")
    print("2 - Positivo / Negativo")
    print("3 - Maior")
    print("4 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
            numero = int(input("Digite um número para verificar se é par ou ímpar: "))
            if numero % 2 == 0:
                print("-------------------------------------------------------")
                print(f"O número {numero} é PAR.")
                print("-------------------------------------------------------")
            else:
                print("-------------------------------------------------------")
                print(f"O número {numero} é ÍMPAR.")
                print("-------------------------------------------------------")
    elif opcao == 2:
            numero = float(input("Digite um número para verificar se é positivo ou negativo: "))
            if numero > 0:
                print("-------------------------------------------------------")
                print(f"O número {numero} é POSITIVO.")
                print("-------------------------------------------------------")
            elif numero < 0:
                print("-------------------------------------------------------")
                print(f"O número {numero} é NEGATIVO.")
            else:
                print("-------------------------------------------------------")
                print("O número é ZERO.")
                print("-------------------------------------------------------")
    elif opcao == 3:
            num1 = float(input("Digite o primeiro número: "))
            num2 = float(input("Digite o segundo número: "))
            if num1 > num2:
                print("-------------------------------------------------------")
                print(f"O maior número é: {num1}")
                print("-------------------------------------------------------")
            elif num2 > num1:
                print("-------------------------------------------------------")
                print(f"O maior número é: {num2}")
                print("-------------------------------------------------------")
            else:
                print("-------------------------------------------------------")
                print("Os números são iguais.")
                print("-------------------------------------------------------")

    elif opcao == 4:
        print("-------------------------------------------------------")
        print("Saindo do programa. Até mais!")
        print("-------------------------------------------------------")
except:
  print("-------------------------------------------------------")
  print("Opção inválida. Por favor, escolha uma opção de 1 a 4.")
  print("-------------------------------------------------------")
```
