```python
# Exercício 1

nome = str(input("Digite seu nome: "))
idade = int(input("Digite sua idade: "))
altura = float(input("Digite sua altura: "))
nota = float(input("Digite sua maior nota do 2º Bimestre: "))
cidade = str(input("Digite o nome da sua cidade: "))

print(f"----------------------------------------------------------------------")
print(f"Variável nome: Valor = {nome}, Tipo = {type(nome)}")
print(f"Variável idade: Valor = {idade}, Tipo = {type(idade)}")
print(f"Variável altura: Valor = {altura}, Tipo = {type(altura)}")
print(f"Variável nota: Valor = {nota}, Tipo = {type(nota)}")
print(f"Variável cidade: Valor = {cidade}, Tipo = {type(cidade)}")
```

```python
# Exercício 2

nota1 = 7.5
nota2 = 8.0
nota3 = 6.5

media = (nota1 + nota2 + nota3) / 3

print(f"A média é igual a: {media}")
```

```python
# Exercício 3

texto1 = "20"
texto2 = "3.5"
numero = 8

final1 = int(texto1) + 10
print(f"Resultado da conversão 1 igual a: {final1}")

final2 = float(texto2) * 2
print(f"Resultado da conversão 2 igual a: {final2}")

final3 = str(numero) + "anos"
print(f"Resultado da conversão 3 igual a: {final3}")
```

```python
# Exercício 5

idade = 16
nota = 8.5
preco = 45.0

print('idade >= 18:', idade >= 18)
print('nota > 7:', nota > 7)
print('preco <= 50:', preco <= 50)
print('idade == 16:', idade == 16)
print('nota != 10:', nota != 10)
```

```python
# Exercício 6

preco_original = 80.0

desconto = preco_original * 0.10

valor_final = preco_original - desconto

eh_menor = valor_final < 75.0

print("Valor do desconto: R$", desconto)
print("Valor final a pagar: R$", valor_final)
print("O valor final é menor que 75.00?:", eh_menor)
```
