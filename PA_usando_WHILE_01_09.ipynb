```python
# Exercício 1 — Contagem

numero = int(input("Digite um número: "))

contador = 1

while contador <= numero:
  print(contador)

  contador += 1
```

```python
# Exercício 2 — Contagem regressiva

contador = int(input("Digite um número: "))

while contador >= 0:

  print(contador)

  contador -= 1
```

```python
# Exercício 3 — Números pares

numero = 0

while numero <= 20:
  print(numero)
  numero = numero + 2
```

```python
# Exercício 4 — Soma dos números

numero = int(input("Digite um número: "))

contador = 1
soma =  0

while contador <= numero:
  soma = soma + contador
  contador = contador + 1

print("Soma dos númeors:", soma)
```

```python
# Exercício 5 — Tabuada

numero = int(input("Digite um número: "))

contador = 1

while contador <= 10:
    resultado = numero * contador
    print(numero, "x", contador, "=", resultado)
    contador = contador + 1
```

```python
# Exercício 6 — Senha

senha = input("Digite a senha: ")

while senha != "1234":
    print("Senha incorreta!")
    senha = input("Digite a senha: ")

print("Acesso permitido!")
```

```python
# Exercício 7 — Somando até digitar zero

soma = 0
numero = int(input("Digite um número: "))

while numero != 0:
    soma = soma + numero
    numero = int(input("Digite um número: "))

print("Soma:", soma)
```

```python
# Exercício 8 — Maior número

maior = 0
numero = int(input("Digite um número: "))

while numero != 0:

    if numero > maior:
        maior = numero


    numero = int(input("Digite um número: "))

print("Maior número:", maior)
```

```python
# Exercício 9 — Média das notas

contador = 1
soma = 0

while contador <= 5:

    nota = float(input("Digite a nota: "))

    soma = soma + nota
    contador = contador + 1


media = soma / 5
print("Média:", media)
```

```python
# Exercício 10 — Caixa eletrônico

saldo = 1000.0
opcao = 0

while opcao != 4:
    print("--- Caixa Eletrônico ---")
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        deposito = float(input("Valor do depósito: "))
        saldo = saldo + deposito

    elif opcao == 2:
        saque = float(input("Valor do saque: "))
        saldo = saldo - saque

    elif opcao == 3:
        print("Seu saldo atual é: R$", saldo)

    elif opcao == 4:
        print("Saindo... Até logo!")

    else:
        print("Opção inválida! Tente novamente.")

    print()
```
