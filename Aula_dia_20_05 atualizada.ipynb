```python
# Escrever um programa que recebe 4 notas do aluno
# Calcula a média
# E imprime aprovado, recuperação e reprovado

# Regras:
# Usar While
# Nota >= a 7, aprovado
# Nota >= a 5, recuperação
# Nota < que 5, reprovado

soma = 0
bimestre = 1

print("Digite suas notas ABAIXO")
while bimestre <= 4:
    nota = float(input(f"Digite sua nota do {bimestre}º bimestre: "))
    soma = soma + nota
    bimestre = bimestre + 1

media = soma / 4
print(f"A sua média foi {media}")

if media >= 7:
    print("Você foi APROVADO!")
elif media >= 5:
    print("Você está de RECUPERAÇÃO!")
else:
    print("Você está REPROVADO!")
```

```python
# Escrever um programa que recebe 4 notas do aluno
# Calcula a média
# E imprime aprovado, recuperação e reprovado

# Regras:
# Usar While
# Nota >= a 7, aprovado
# Nota >= a 5, recuperação
# Nota < que 5, reprovado

soma_notas = 0.0
contador = 0
print("Digite suas notas ABAIXO")
while contador < 4:
    nota = float(input("Digite a nota do " + str(contador + 1) + "º bimestre: "))
    soma_notas = soma_notas + nota
    contador = contador + 1

media = soma_notas / contador
print("\nA média do aluno é: " + str(media))

if media >= 7:
    print("Você foi aprovado!")
elif media >= 5:
    print("Você está recuperação.")
else:
    print("Você foi reprovado.")
```

```python
# Você foi contratado para desenvolver um sistema simples de caixa eletrônico.

# O sistema deverá funcionar através de um menu interativo utilizando a estrutura de repetição while.

# O programa deve iniciar com um saldo inicial de R$ 1.000,00 e permitir que o usuário escolha diferentes operações até decidir sair do sistema.

# Funcionalidades:

# 1 - ver saldo;
# 2 - Depositar dinheiro
# 3 - sacar dinheiro
# 4 - sair

# Regras:

# ver saldo
# Ao escolher essa opção, o sistema deverá mostrar o saldo atual da conta;

# Depositar dinheiro
# O usuário deverá informar um valor para depósito
# O valor digitado deve ser somado ao saldo.

# Sacar dinheiro
# O usuário deverá informar o valor do saque.

# O sistema devera verificar:
# Se houver suficiente - realizar o saque.
# Se não houver saldo suficiente - mostrar mensagem de "saldo insuficiente".

# Sair

# O programa encerrar apenas quando o usuário escolher a opção.

saldo = 1000.0
opcao = 0

while opcao != 4:
    print("--- CAIXA ELETRÔNICO ---")
    print("1 - Ver Saldo")
    print("2 - Depositar Dinheiro")
    print("3 - Sacar Dinheiro")
    print("4 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        print(f"Seu saldo atual é: R$ {saldo}")

    elif opcao == 2:
        deposito = float(input("Digite o valor do depósito: "))
        saldo = saldo + deposito
        print(f"Depósito de R$ {deposito} realizado com sucesso!")

    elif opcao == 3:
        saque = float(input("Digite o valor do saque: "))
        if saque <= saldo:
            saldo = saldo - saque
            print(f"Saque de R$ {saque} realizado!")
        else:
            print("Saldo insuficiente!")

    elif opcao == 4:
        print("Encerrando o sistema. Até logo!")

    else:
        print("Opção inválida! Tente novamente.")
```
