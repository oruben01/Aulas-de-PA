```python
# Jogo Menu inteligente

# Criar um sistema que funcione até o usuário escolher sair.

# Menu

# 1 - Par/ Impar
# 2 - Positivo / Negativo
# 3 - Maior
# 4 - Sair

opcao = ""

try:

  while opcao != 4:

    print("--- Menu ---")
    print("1 - Par/ Impar")
    print("2 - Positivo / Negativo")
    print("3 - Maior")
    print("4 - Sair")

    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
            numero = int(input("Digite um número para verificar se é par ou ímpar: "))
            if numero % 2 == 0:
                print("-------------------------------------------------------")
                print(f"O número {numero} é PAR.")
                print("-------------------------------------------------------")
            else:
                print("-------------------------------------------------------")
                print(f"O número {numero} é ÍMPAR.")
                print("-------------------------------------------------------")
    elif opcao == 2:
            numero = float(input("Digite um número para verificar se é positivo ou negativo: "))
            if numero > 0:
                print("-------------------------------------------------------")
                print(f"O número {numero} é POSITIVO.")
                print("-------------------------------------------------------")
            elif numero < 0:
                print("-------------------------------------------------------")
                print(f"O número {numero} é NEGATIVO.")
            else:
                print("-------------------------------------------------------")
                print("O número é ZERO.")
                print("-------------------------------------------------------")
    elif opcao == 3:
            num1 = float(input("Digite o primeiro número: "))
            num2 = float(input("Digite o segundo número: "))
            if num1 > num2:
                print("-------------------------------------------------------")
                print(f"O maior número é: {num1}")
                print("-------------------------------------------------------")
            elif num2 > num1:
                print("-------------------------------------------------------")
                print(f"O maior número é: {num2}")
                print("-------------------------------------------------------")
            else:
                print("-------------------------------------------------------")
                print("Os números são iguais.")
                print("-------------------------------------------------------")

    elif opcao == 4:
        print("-------------------------------------------------------")
        print("Saindo do programa. Até mais!")
        print("-------------------------------------------------------")
except:
  print("-------------------------------------------------------")
  print("Opção inválida. Por favor, escolha uma opção de 1 a 4.")
  print("-------------------------------------------------------")
```

```python
# Urna Eletrônica

# Criar um programa de votação com 3 candidatos

# Votar candidato A
# Votar candidato B
# Votar candidato C
# Votar Branco
# Votar Nulo
# Encerrar votação

# O sistema deve repetir até encerrar a votação
# Contar votos de cada candidato
# Contar votos brancos e nulos
# Ao encerra, mostrar:

# Total de Votos
# Total de votos por candidato
# Total votos brancos
# Total votos nulos
# Vencedor da eleição

# Regras extras

# Se houver empate entre candidatos : Mostrar eleição empatada
# Se não houver votos válidos, mostrar : "nenhum candidato recebeu votos."

  votos_a = 0
votos_b = 0
votos_c = 0
votos_branco = 0
votos_nulo = 0

print("=== INÍCIO DO SISTEMA DE VOTAÇÃO ===")

while True:
    print()
    print("OPÇÕES DE VOTO:")
    print("1 - Candidato A")
    print("2 - Candidato B")
    print("3 - Candidato C")
    print("4 - Voto Branco")
    print("5 - Voto Nulo")
    print("6 - Encerrar Votação")

    opcao = int(input("Digite o número da sua opção: "))

    if opcao == 1:
        votos_a = votos_a + 1
        print("Voto computado para Candidato A.")
    elif opcao == 2:
        votos_b = votos_b + 1
        print("Voto computado para Candidato B.")
    elif opcao == 3:
        votos_c = votos_c + 1
        print("Voto computado para Candidato C.")
    elif opcao == 4:
        votos_branco = votos_branco + 1
        print("Voto Branco computado.")
    elif opcao == 5:
        votos_nulo = votos_nulo + 1
        print("Voto Nulo computado.")
    elif opcao == 6:
        print("Encerrando a votação...")
        break
    else:
        print("Opção inválida! Por favor, digite um número de 1 a 6.")

total_votos = votos_a + votos_b + votos_c + votos_branco + votos_nulo
votos_validos = votos_a + votos_b + votos_c

print()
print("==============================")
print("     RESULTADO DA ELEIÇÃO     ")
print("==============================")
print(f"Total de Votos: {total_votos}")
print(f"Total Candidato A: {votos_a}")
print(f"Total Candidato B: {votos_b}")
print(f"Total Candidato C: {votos_c}")
print(f"Total Votos Brancos: {votos_branco}")
print(f"Total Votos Nulos: {votos_nulo}")
print("------------------------------")

if votos_validos == 0:
    print("Resultado: Nenhum candidato recebeu votos.")
else:
    if votos_a > votos_b and votos_a > votos_c:
        print("Vencedor da eleição: Candidato A")
    elif votos_b > votos_a and votos_b > votos_c:
        print("Vencedor da eleição: Candidato B")
    elif votos_c > votos_a and votos_c > votos_b:
        print("Vencedor da eleição: Candidato C")

    elif votos_a == votos_b and votos_a == votos_c:
        print("Resultado: Segundo turno entre Candidatos A, B e C (Empate Triplo).")
    elif votos_a == votos_b and votos_a > votos_c:
        print("Resultado: Segundo turno entre Candidato A e Candidato B.")
    elif votos_a == votos_c and votos_a > votos_b:
        print("Resultado: Segundo turno entre Candidato A e Candidato C.")
    elif votos_b == votos_c and votos_b > votos_a:
        print("Resultado: Segundo turno entre Candidato B e Candidato C.")

print("==============================")
```
