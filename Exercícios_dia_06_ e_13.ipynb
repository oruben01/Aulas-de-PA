```python
# 1 - Contagem simples: O usuário digita um número e o programa imprime do zero até o número.

# Entrada

num = int(input("Digite um número: "))

# Processamento

for i in range (0, num + 1):
  print(i)
```

```python
# 2 - Contagem regressiva: o usuário digita um número e o programa imprime do número até o zero.

# Entrada

num = int(input("Digite um número: "))
for i in range (num,- 1, -1):
  print(i)
```

```python
# 3 - Números pares: o usuário digita um número e o programa imprimi os números pares.

# Entrada

num = int(input("Digite um número: "))

# Processamento e Saída

for i in range (0, num + 1, 2):
  print(i)
```

```python
# 4 - Números ímpares: o usuário digita um número e o programa imprimi os números ímpares.

# Entrada

num = int(input("Digite um número: "))

# Processamento

for i in range (1, num + 1, 2):
  print(i)
```

```python
# 5 - Soma: o usuário digita um número e o programa soma os números do zero até o número.

# Entrada

numero = int(input("Digite um número para somar do zero até ele: "))

soma = 0
for i in range(numero + 1):
    soma += i

print(f"A soma dos números de 0 até {numero} é: {soma}")
```

```python
# 6 - Tabuada: o usuário digita um número e o programa imprime a tabuada do número.

# Entrada

numero = int(input("Digite um número para ver a tabuada: "))

print(f"Tabuada do {numero}:")
for i in range(1, 11):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```

```python
# 8 - Contar vogais: o usuário digita uma palavra e o programa conta as vogais.

# Entrada

palavra = str(input("Digite uma palavra: "))
vogais = "aeiou"
letra = 0

for i in (palavra):
  if i in vogais:
    letra += 1

print(f"A palavra tem {letra} vogais.")
```

```python
# 9 - Maior número digitado: o usuário digita quatro números e o programa imprime o maior número digitado.

# Entrada

num1 = float(input("Digite o 1º número: "))
maior_numero = num1

num2 = float(input("Digite o 2º número: "))
if num2 > maior_numero:
    maior_numero = num2

num3 = float(input("Digite o 3º número: "))
if num3 > maior_numero:
    maior_numero = num3

num4 = float(input("Digite o 4º número: "))
if num4 > maior_numero:
    maior_numero = num4

print(f"O maior número digitado é: {maior_numero}")
```

```python
# 10 - O usuário digita um número entre 01 e 100 e o programa mostra quais são divisíveis por 3 e 5.
# PROFESSOR, FIQUEI MUITO EM DÚVIDA NESSE CÓDIGO

# Entrada

numero = int(input("Digite um número entre 1 e 100 (assume-se entrada válida): "))

print(f"Números entre 1 e {numero} divisíveis por 3 e 5:")
found_any = False

for i in range(1, numero + 1):
    if i % 3 == 0 and i % 5 == 0:
        print(i)
        found_any = True

if not found_any:
    print(f"Não há números entre 1 e {numero} que sejam divisíveis por 3 e 5.")
```
