```python
# Exercício 1 – Idade para dirigir

idade = int(input("Digite a sua idade: "))

print("-----------------------------------------------------------------------")
if idade >= 18:
    print("Você PODE tirar a carteira de motorista!")
else:
    print("Você ainda NÃO PODE tirar a carteira de motorista.")
```

```python
# Exercício 2 - Par ou Ímpar

numero = int(input("Digite um número: "))

if numero % 2 == 0:
    print("O número é PAR!")
else:
    print("O número é ÍMPAR!")
```

```python
# Exercício 3 – Maior de dois números

num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

print("-----------------------------------------------------------------------")
if num1 > num2:
    print(f"O primeiro número ({num1}) é o maior!")
else:
    if num2 > num1:
        print(f"O segundo número ({num2}) é o maior!")
    else:
        print("Os dois números são IGUAIS!")
```

```python
# Exercício 4 – Situação do aluno


nota = float(input("Digite a nota do aluno: "))

print("-----------------------------------------------------------------------")
if nota >= 7:
    print("APROVADO")
else:
    if nota >= 5:
        print("RECUPERAÇÃO")
    else:
        print("REPROVADO")
```

```python
# Exercício 5 - Desconto na compra

valor_compra = float(input("Digite o valor da compra: R$ "))

if valor_compra >= 100:
    desconto = valor_compra * 0.10
    valor_final = valor_compra - desconto
    print(f"Desconto de 10% aplicado! Valor final: R$ {valor_final:.2f}")
else:
    print(f"Sem desconto. Valor final: R$ {valor_compra:.2f}")
```

```python
# Exercício 6 - Positivo, Negativo ou Zero

numero = float(input("Digite um número para começar a verificação: "))

print("-----------------------------------------------------------------------")
if numero > 0:
  print("O número é POSITIVO!")
else:
  if numero < 0:
    print("O número é NEGATIVO!")
  else:
    print("O número é ZERO")
```

```python
# Exercício 7 – Classificação de idade

idade = int(input("Digite a sua idade: "))

print("-----------------------------------------------------------------------")
if idade < 12:
  print("Classificação: CRIANÇA")
else:
  if idade <= 17:
    print("Classificação: ADOLESCENTE")
  else:
    if idade <= 59:
      print("Classificação: ADULTO")
    else:
      print("Classificação: IDOSO")
```

```python
# Exercício 8 – Média de duas notas

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))


media = (nota1 + nota2) / 2

print("-----------------------------------------------------------------------")
if media >= 7:
  print(f"Média: {media: .1f} - APROVADO!")
else:
  print(f"Média: {media: .1f} - REPROVADO!")
```

```python
# Exercício 9 – Login simples

usuario = input("Usuário: ")
senha = input("Senha: ")

print("-----------------------------------------------------------------------")
if usuario == "RUBINHO" and senha == "1234":
  print("Login realizado com sucesso!")
else: print("Usuário ou senha incorretos.")
```

```python
# Exercício 10 – Calculadora de IMC

peso = float(input("Digite o seu peso(em KG): "))
altura = float(input("Digite a sua altura (em metros): "))

imc = peso / (altura * altura)

print("-----------------------------------------------------------------------")
print(f"Seu IMC é: {imc: .2f}")

if imc < 18.5:
  print("ABAIXO do peso")
else:
  if imc <= 24.9:
    print("Peso NORMAL")
  else:
    if imc <= 29.9:
      print("SOBREPESO")
    else:
      print("OBESIDADE")
```
