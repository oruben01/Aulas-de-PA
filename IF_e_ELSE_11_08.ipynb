```python
# Exercício 1 – Idade para dirigir

idade = int(input("Digite a sua idade: "))

print("-----------------------------------------------------------------------")
if idade >= 18:
    print("Você PODE tirar a carteira de motorista!")
else:
    print("Você ainda NÃO PODE tirar a carteira de motorista.")
```

```python
# Exercício 2 - Par ou Ímpar

numero = int(input("Digite um número: "))

if numero % 2 == 0:
    print("O número é PAR!")
else:
    print("O número é ÍMPAR!")
```

```python
# Exercício 3 – Maior de dois números

num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

print("-----------------------------------------------------------------------")
if num1 > num2:
    print(f"O primeiro número ({num1}) é o maior!")
else:
    if num2 > num1:
        print(f"O segundo número ({num2}) é o maior!")
    else:
        print("Os dois números são IGUAIS!")
```

```python
# Exercício 4 – Situação do aluno


nota = float(input("Digite a nota do aluno: "))

if nota >= 7:
    print("Aprovado")
else:
    if nota >= 5:
        print("Recuperação")
    else:
        print("Reprovado")
```
